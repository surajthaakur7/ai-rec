```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
results = pd.DataFrame({
    'Model': ['Collaborative Filtering', 'Content Based Filtering', 'Hybrid Model'],
    'MAE':  [2.51, 0.83, 0.87],
    'MSE':  [8.88, 1.45, 1.15],
    'R²':   [-5.0, -0.45, 0.14]
})

print("=== Final Model Comparison ===")
print(results.to_string(index=False))
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Model', y='MAE', data=results, palette='viridis')
plt.title('MAE Comparison Across Models')
plt.ylabel('MAE (Lower is Better)')
plt.xticks(rotation=15)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Model', y='MSE', data=results, palette='viridis')
plt.title('MSE Comparison Across Models')
plt.ylabel('MSE (Lower is Better)')
plt.xticks(rotation=15)
plt.show()
```

```python
plt.figure(figsize=(8,5))
colors = ['red' if r < 0 else 'green' for r in results['R²']]
plt.bar(results['Model'], results['R²'], color=colors)
plt.title('R² Comparison Across Models')
plt.ylabel('R² (Higher is Better)')
plt.xticks(rotation=15)
plt.axhline(y=0, color='black', linestyle='--')
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# MAE
axes[0].bar(results['Model'], results['MAE'], color='steelblue')
axes[0].set_title('MAE (Lower is Better)')
axes[0].set_xticklabels(results['Model'], rotation=15)

# MSE
axes[1].bar(results['Model'], results['MSE'], color='orange')
axes[1].set_title('MSE (Lower is Better)')
axes[1].set_xticklabels(results['Model'], rotation=15)

# R²
colors = ['red' if r < 0 else 'green' for r in results['R²']]
axes[2].bar(results['Model'], results['R²'], color=colors)
axes[2].set_title('R² (Higher is Better)')
axes[2].set_xticklabels(results['Model'], rotation=15)
axes[2].axhline(y=0, color='black', linestyle='--')

plt.suptitle('Complete Model Evaluation', fontsize=14)
plt.tight_layout()
plt.show()
```
